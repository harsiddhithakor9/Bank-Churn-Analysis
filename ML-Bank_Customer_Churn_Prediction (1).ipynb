```python
#Q1. Load the cleaned Bank Customer dataset. Identify the Type of 
#ML Problem.  
#Select 'Churn' as target variable, and print the shape of X and y.  
```

```python
x=df.drop(columns=["customer_id","churn"])
```

```python
y=df["churn"]
```

```python
x.shape
```

```python
y.shape
```

```python
#Q2. Identify categorical columns, apply Encoding techniques (if 
#necessary, perform Standard Scaling on numerical features.
```

```python
encoded=pd.get_dummies(x,columns=["country","gender"],drop_first=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x_train,x_test,y_train,y_test=train_test_split(encoded,y,test_size=0.2,random_state=42)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(x_train)
```

```python
X_test_scaled = scaler.transform(x_test)
```

```python
#Q.3.Plot churn distribution. Analyze the relationship between 
#CreditScore, Age, and Balance with churn
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
sns.countplot(data=df, x='churn', palette='Set2')
plt.title('Churn Distribution')

plt.subplot(2, 2, 2)
sns.boxplot(data=df, x='churn', y='credit_score', palette='Set2')
plt.title('CreditScore vs Churn')

plt.subplot(2, 2, 3)
sns.boxplot(data=df, x='churn', y='age', palette='Set2')
plt.title('Age vs Churn')

plt.subplot(2, 2, 4)
sns.boxplot(data=df, x='churn', y='balance', palette='Set2')
plt.title('Balance vs Churn')

plt.show()
```

```python
#Q.4 Split the dataset into training and testing sets. Train a SVM model 
#and display accuracy Score, Confusion Matrix, Classification  
#Report..  
```

```python
from sklearn.svm import SVC
```

```python
svm_model = SVC(random_state=42)
```

```python
svm_model.fit(X_train_scaled, y_train)
```

```python
X_test_scaled = scaler.transform(x_test)
```

```python
svm_model.score(x_train,y_train)
```

```python
svm_model.score(x_test,y_test)
```

```python
y_pred_svm = svm_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_test,y_pred_svm)
```

```python
confusion_matrix(y_test,y_pred_svm)
```

```python
classification_report(y_test,y_pred_svm)
```

```python
cm_svm = confusion_matrix(y_test, y_pred_svm)
sns.heatmap(cm_svm, annot=True, fmt='d', cmap='Blues')
plt.title('SVM Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
