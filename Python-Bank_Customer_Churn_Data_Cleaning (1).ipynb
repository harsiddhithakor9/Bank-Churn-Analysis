```python
#1.Fill all missing values in the Country column with the name of the 
#country in which you are currently residing. 
```

```python
df["country"]=df["country"].fillna("India")
```

```python
#2.Identify the total number of female records in the dataset. Compare the 
#count of males and females, and fill all missing values in the Gender 
#column with the gender that has the higher count.
```

```python
female_count=(df["gender"]=="Female").sum()
```

```python
male_count=(df["gender"]=="Male").sum()
```

```python
higher_gender="Male" if male_count>female_count else "Female"
df["gender"]=df["gender"].fillna(higher_gender)
```

```python
#3.Analyze the Age column, calculate the average age, and fill all 
#missing age values using this average. Ensure that the age values are 
#stored as integers (no decimal values).
```

```python
avg_age=round(df["age"].mean())
```

```python
df["age"]=df["age"].fillna(avg_age).astype(int)
```

```python
df
```

```python
#4.Handle missing values in the Balance column as follows: 
#Fill the first 500 missing values with the average balance, 
#Fill the next 500 missing values with the maximum balance, 
#Fill the remaining missing values with 0. 
#Ensure the balance values contain exactly 2 decimal places.
```

```python
avg_balance=round(df["balance"].mean(),2)
```

```python
max_balance=round(df["balance"].max(),2)
```

```python
null_balance=df[df["balance"].isnull()].index
```

```python
# First 500 -> Average balance
```

```python
df.loc[null_balance[:500],"balance"]=avg_balance
```

```python
df.loc[null_balance[500:1000],"balance"]=max_balance
```

```python
# Remaining -> 0
```

```python
df.loc[null_balance[1000:],"balance"]=0.00
```

```python
df["balance"]=df["balance"].round(2)
```

```python
df
```

```python
#5.For the Active_Member column, fill missing values such that 50% of 
#them are assigned 1 (active) and the remaining 50% are assigned 0 
#(inactive).
```

```python
null_active=df[df["active_member"].isnull()].index
```

```python
half=len(null_active)//2
```

```python
df.loc[null_active[:half],"active_member"]=1
```

```python
df.loc[null_active[half:],"active_member"]=0
```

```python
df["active_member"]=df["active_member"].astype(int)
```

```python
df
```

```python
#6.Calculate the average value of the Estimated_Salary column and fill 
#all missing salary values with this average, ensuring the values have 
#exactly 2 decimal places.
```

```python
avg_salary=round(df["estimated_salary"].mean(),2)
```

```python
df["estimated_salary"]=df["estimated_salary"].fillna(avg_salary).round(2)
```
